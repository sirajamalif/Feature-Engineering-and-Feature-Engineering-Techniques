# Euclidean Distance

```python
import numpy as np

def euclidean_distance(a, b):
    a = np.array(a)
    b = np.array(b)
    return np.sqrt(np.sum((a - b) ** 2))


point1 = [1, 2]
point2 = [4, 6]
distance = euclidean_distance(point1, point2)
print(f"Euclidean Distance: {distance}")
```

```python
from sklearn.metrics.pairwise import euclidean_distances
import numpy as np

a = np.array([[1, 2]])
b = np.array([[4, 6]])

distance = euclidean_distances(a, b)
print(f"Euclidean Distance: {distance[0][0]}")
```

# Manhattan Distance

```python
import numpy as np

def manhattan_distance(a, b):
    a = np.array(a)
    b = np.array(b)
    return np.sum(np.abs(a - b))


point1 = [1, 2]
point2 = [4, 6]
distance = manhattan_distance(point1, point2)
print(f"Manhattan Distance: {distance}")
```

```python
from sklearn.metrics.pairwise import manhattan_distances
import numpy as np

a = np.array([[1, 2]])
b = np.array([[4, 6]])

distance = manhattan_distances(a, b)
print(f"Manhattan Distance: {distance[0][0]}")
```

# Minkowski Distance

```python
import numpy as np

def minkowski_distance(a, b, p):
    a = np.array(a)
    b = np.array(b)
    return np.sum(np.abs(a - b) ** p) ** (1 / p)


point1 = [1, 2]
point2 = [4, 6]

# Try with p=1 (Manhattan), p=2 (Euclidean), or any other positive value, P cannot be 0 because 1/P = Undefined
distance = minkowski_distance(point1, point2, p=3)
print(f"Minkowski Distance (p=3): {distance}")
```

# Hamming Distance

```python
import numpy as np

def hamming_distance(a, b):
    a = np.array(a)
    b = np.array(b)
    if a.shape != b.shape:
        raise ValueError("Input arrays must have the same shape")
    return np.sum(a != b)

a = [1, 0, 1, 1]
b = [1, 1, 0, 1]
distance = hamming_distance(a, b)
print(f"Hamming Distance: {distance}")
```

```python
from sklearn.metrics.pairwise import pairwise_distances
import numpy as np

a = np.array([[1, 0, 1, 1]])
b = np.array([[1, 1, 0, 1]])


distance = pairwise_distances(a, b, metric='hamming')
print(f"Hamming Distance: {distance[0][0]}")
raw_distance = distance[0][0] * a.shape[1]
print(f"Raw Distance: {raw_distance}")
```
